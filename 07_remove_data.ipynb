We can remove the rows and columns from our data frames.

2. Removing multiple columns

### 1. Removing a single column
---

A single column can be removed from the dataframe using the **`drop`** method.

`Syntax:` df.drop[columns = [column name(s) ] inplace = True]

*inplace=True* is also used because it modifies the original DataFrame directly instead of creating a copy.

```python
import pandas as pd

data = {
    "Name": ["Alex", "Finch", "Root", "Williamson", "Inzamam", "Ponting", "Kohli"],
    "Age": [23, 21, 34, 32, 29, 45, 32],
    "Salary": [32000, 50000, 58000, 21000, 88000, 43000, 34000],
    "Performance Score": [45, 68, 90, 79, 53, 75, 90]
}

df = pd.DataFrame(data)
print(f"Original data:\n\n{df}")

# removing Salary column
df = df.drop(columns = ['Salary'])
print(f"\n\nData after removing Salary column:\n\n{df}")
```
