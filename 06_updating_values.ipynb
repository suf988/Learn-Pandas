# Updating the values
---

1. Updating value in specific cell
2. Updating values of different columns in a single row
3. Updating value in multiple cells (columns)

### 1. Updating value in specific cell
---

We use the function `.loc` to update a single cell.

`Syntax:` df.loc["row index", "column name"] = new value

```python
import pandas as pd

data = {
    "Name": ["Alex", "Finch", "Root", "Williamson", "Inzamam", "Ponting", "Kohli"],
    "Age": [23, 21, 34, 32, 29, 45, 32],
    "Salary": [32000, 50000, 58000, 21000, 88000, 43000, 34000],
    "Performance Score": [45, 68, 90, 79, 53, 75, 90]
}

df = pd.DataFrame(data)
print(f"Original data:\n\n{df}")

# updating the salary of row no. 4 from 88000 to 125000

df.loc[4,"Salary"] = 125000

print(f"Updated data:\n\n{df}")
```

### 2. Updating values of different columns in a single row
---

We can update multiple columns of a single row

`Example: Changing salary and age of row 5:`

`Syntax:` df.loc[row, [column 1, column 2] ] = new value

```python
df.loc[5, ['Salary', 'Age'] ] = [100000, 47]

print(f"New data after updating Age and Salary of row 5:\n\n{df}")
```

### 3. Updating value in multiple rows
---

We can update multiple values in a single column.
Like we will update the salaries by 10%

`Syntax:` df['column name'] = updated value

```python
df['Salary'] = df['Salary'] * 1.10
print(f"Updating multiple rows of same column:\n\n{df}")

```
